```python
# Relational / Comparison Operators in python. Relational operators are used to compare two values.
# The results of every relational operator is always a boolean value: True or False
x = 10
y = 20

print(x==y) # Equal to == operator checks whether two values are equal.
print(x>y) # Greater Than > operator checks whether the left-hand value is greater than the right-hand value.
print(x<y)# Less Than < operator checks whether the left-hand value is less than the right-hand value.
print(x>=y)# Greater Than Equal To >= operator checks two conditions:
                                            # Left value is greater than the right value 
                                            # OR both values are equal
print(x<=y) # Less Than or Equal to <= operator checks two conditions:
                                        # Left value is less than the right value
                                        # OR both values are equal
print(x!=y)# Not Equal To != operator checks whether two values are different.
```

```python
# Logical and Operator in Python:
# The and operator is used to combine two or more conditions. and = BOTH conditions must be True.
x = 10
y = 20
print(x<y and y>x)
print(x<y and x==y)
print(x>y and x<y)
print(x>y and x==y)
```

```python
x = [10,20] # Python  creates a list object in memory.
y = x # python does not create another list. Instead, y refers to the same list object.
# So, both x and y point to the same object.
print(x is y)# True Why? Because x and y refer to the same list  object.
# is Operators: The is operator checks whether two variables refers to the same object in memory.
print(id(x))
print(id(y))
print(x==y) # == Operators checks value / content equality.
```

```python
# == vs is with List
x = [10,20] # python creates a list object in memory and x refers to that object.
y = [10,20] # python creates another separate list object. Although the values are the same, the two lists are different objects
            # Because list belongs to mutable Type.
print(x==y)# Output: True   Why? The == operator comapres the values/content of two objects.
print(x is y) # Output: False Why? The is operator checks object identify.
x.append(33) # The list referred to by x is modified. Only the x list changes because x and y refer to different list objects.
print(x==y)# Output: False Because: x -> [10,20,33] y -> [10,20] Their contents are now different.
print(id(x))
print(id(y))
```

```python
# Exercise 4: ATM Difficult challenges

correct_pin = 1234

pin = int(input("Enter your PIN: "))

if pin != correct_pin:
    print("Incorrect PIN.")
    
else:
    account_active = input(
        "Is your account active? (yes/no): "
    ).lower()

    if account_active != "yes":
        print("Account is inactive.")

    else:
        balance = float(input("Enter account balance: ₹"))
        amount = float(input("Enter withdrawal amount: ₹"))

        # Check amount positive
        if amount <= 0:
            print("Withdrawal amount must be positive.")

        # Check amount multiple of 100
        elif amount % 100 != 0:
            print("Withdrawal amount must be a multiple of ₹100.")

        # Check sufficient balance and minimum ₹500 remaining
        elif amount > balance:
            print("Insufficient balance.")

        elif balance - amount < 500:
            print("Withdrawal failed.")
            print("Minimum ₹500 balance must remain in your account.")

        # Extra verification for amount > ₹20,000
        elif amount > 20000:
            verification = input(
                "Extra verification required. Enter 'yes' to verify: "
            ).lower()

            if verification == "yes":
                balance = balance - amount

                print("Withdrawal successful!")
                print("Withdrawn Amount: ₹", amount)
                print("Remaining Balance: ₹", balance)

            else:
                print("Extra verification failed.")

        else:
            balance = balance - amount

            print("Withdrawal successful!")
            print("Withdrawn Amount: ₹", amount)
            print("Remaining Balance: ₹", balance)
```

```python
import builtins
def calculate_avg(marks):
    if len(marks) == 0:
        print("No marks provided")
        return

    for mark in marks:
        if mark < 0 or mark > 100:
            print("Invalid marks: ",mark)
            return

    total = builtins.sum(marks)
    avg = total/len(marks)

    print("Subject Marks:",marks)
    print("Total Marks:",total)
    print("Average Marks:",avg)

calculate_avg([80,75,90,85])
calculate_avg([70,110,60])
calculate_avg([])
```

```python
# Student records 
students = [
    {"name" : "Aman" , "marks" : [80, 75, 90]},
    {"name" : "Riya" , "marks" : [35, 40, 30]},
    {"name" : "Rahul" , "marks" : [90, 85, 95]},
    {"name" : "Neha" , "marks" : [60, 70, 65]}
]
def generate_result(student_record):
    for student in (student_record):
        name=student['name'] #Aman
        marks=student['marks'] #[80, 75, 90]

        avg=sum(marks) /len(marks) #
        if avg>=40:
            status="Pass"
        else:
            status="Fail"

        yield {
            "name":name,
            "average":avg,
            "status": status
        }

ex = generate_result(students)
print(next(ex))
print(next(ex))
print(next(ex))
print(next(ex))
```
